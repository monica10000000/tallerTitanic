1. ¿Cuántos pasajeros hay en el dataset y cuántos PassengerId distintos aparecen? ¿Coinciden los dos resultados?

```python
totalPasajeros = len (df)
idsDistintos = df ['PassengerId'].nunique ()
coinciden = "Sí" if totalPasajeros == idsDistintos else "No"

tabla1 = pd.DataFrame ({
    'Total de pasajeros': [totalPasajeros], 'PassengerId distintos': [idsDistintos],
    '¿Coinciden?': [coinciden]
})
tabla1
```

2. ¿Cuántos pasajeros de cada clase aparecen en el archivo? ¿Qué clase tiene más pasajeros registrados?

```python
print ('Hay', len (df[df ['Pclass']== 1]), 'pasajeros en primera clase')
print ('Hay', len (df[df ['Pclass']== 2]), 'pasajeros en segunda clase')
print ('Hay', len (df[df ['Pclass']== 3]), 'pasajeros en tercera clase')
```

```python
primeraClse = len(df[df ['Pclass']==1])
segundaClse = len(df[df ['Pclass']==2])
terceraClse = len(df[df ['Pclass']==3])

if primeraClse > segundaClse and primeraClse > terceraClse:
    print (f"La primera clase tiene mas pasajeros {primeraClse}")
elif segundaClse > primeraClse and segundaClse > terceraClse:
    print (f"La segunda clase tiene mas pasajeros {segundaClse}")
elif terceraClse > primeraClse and terceraClse > segundaClse:
    print (f"La tercera clase tiene mas pasajeros {terceraClse}")
else:
    print("Hay un empate entre las clases")
```

3. ¿Cuántos pasajeros embarcaron en cada puerto y cuántos tienen el puerto sin registrar?

```python
cherburgo = len (df [df ['Embarked'] == 'C'])
queenstown = len (df [df ['Embarked'] == 'Q'])
southampton = len (df [df ['Embarked'] == 'S'])
sinRegistrar = len (df) - (cherburgo + queenstown + southampton)

tabla3 = pd.DataFrame ({ 'Cherburgo' : [cherburgo], 'Queenstown': [queenstown], 'Southampton' : [southampton], 'Sin registar' : [sinRegistrar]
})
tabla3
```

4. ¿Cuál es la tarifa promedio, mínima y máxima registrada para cada clase?

```python
tabla4 = df.groupby ('Pclass') ['Fare'].agg (['mean', 'min','max']).round(2).reset_index()
tabla4.columns = ['Clase', 'Promedio', 'Tarifa minima', 'Tarifa maxima']
tabla4
```

5. ¿Cuántos pasajeros de cada clase embarcaron en cada puerto?

```python
clase1 = df[ df['Pclass'] ==1]
clase2 = df[ df['Pclass'] ==2]
clase3 = df[ df['Pclass'] ==3]

puerto1 = clase1 ['Embarked'].fillna ('Sin registrar').value_counts()
puerto2 = clase2 ['Embarked'].fillna ('Sin registrar').value_counts()
puerto3 = clase3 ['Embarked'].fillna ('Sin registrar').value_counts()

tabla5 = pd.DataFrame ({
    'Puerto 1' : puerto1, 'Puerto 2' : puerto2, 'Puerto 3' : puerto3
})
tabla5
```

9. ¿Cuántos códigos de boleto distintos hay? ¿Cuántos aparecen asociados a más de un pasajero y cuál aparece en más registros?

```python
boletosDistintos = df ['Ticket'].nunique()
conteoBoletos = df ['Ticket'].value_counts()

boletoMas1Pasajero = len (conteoBoletos [ conteoBoletos >1])

masRegistros = df['Ticket'].mode ()
tabla10 = pd.DataFrame ({
    'Boletos distintos': [boletosDistintos],
    'Con más de 1 pasajero': [boletoMas1Pasajero],
    'Boleto más repetido': [masRegistros],
})
tabla10
```
